```python
import cv2
import numpy as np
import matplotlib.pyplot as plt
```

```python
image = cv2.imread("images/j.jpeg", 0) 
plt.imshow(image)
```

```python
kernel = np.ones((5,5),np.uint8)

# Dilate the image
dilation = cv2.dilate(image, kernel, iterations = 4)
plt.imshow(dilation)
```

```python
# Erode the image
erosion = cv2.erode(image, kernel, iterations = 1)
plt.imshow(erosion)
```

```python
image_o_noise = cv2.imread("images/i_outside_noise.jpeg", 0) 
image_o_noise = cv2.cvtColor(image_o_noise, cv2.COLOR_BGR2RGB)
plt.imshow(image_o_noise)
```

```python
opening = cv2.morphologyEx(image_o_noise, cv2.MORPH_OPEN, kernel)
plt.imshow(opening)
```

### Closing

```python
# Reads in a binary image
j_inside_noise = cv2.imread("images/j_inside_noise.jpeg", 0) 
j_inside_noise = cv2.cvtColor(j_inside_noise, cv2.COLOR_BGR2RGB)
plt.imshow(j_inside_noise)
```

```python
closing = cv2.morphologyEx(j_inside_noise, cv2.MORPH_CLOSE, kernel)
plt.imshow(closing)
```
